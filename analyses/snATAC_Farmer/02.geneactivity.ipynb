## Peak calling on concatenated dataset

```python
fileName = 'combined'
```

```python
import scanpy as sc
```

```python
data = sc.read(filename = (saveDIR + fileName + '_2.filteringMADs.h5ad'))
```

```python
data
```

```python
data2 = sc.read(filename = (saveDIR + fileName + '_3.features.h5ad'))
```

```python
data2
```

```python
subset = data[data2.obs_names,]
```

```python
subset.obsm['X_umap'] = data2.obsm['X_umap']
```

```python
subset.obs['leiden'] = data2.obs['leiden']
```

```python
subset.write(filename = (saveDIR + fileName + '_2.filteringMADs.ForGeneActivity.h5ad'))
```

```python
data = snap.read(filename = (saveDIR + fileName + '_2.filteringMADs.ForGeneActivity.h5ad'), backed='r+', backend=None)
```

```python
data
```

```python
gene_matrix = snap.pp.make_gene_matrix(data, gene_anno = gtf,
                                       upstream = 500,
                                       counting_strategy = 'paired-insertion',
                                       gene_name_key = 'gene_id')
```

```python
gene_matrix
```

```python
gene_matrix.obsm['X_umap'] = data2.obsm['X_umap']
```

```python
gene_matrix.obs['leiden'] = data2.obs['leiden']
```

```python
gene_matrix.obsm['X_spectral'] = data2.obsm['X_spectral']
```

```python
data.close()
```

```python
gene_matrix.write(filename = (saveDIR + fileName + '_4.geneMtx.h5ad'))
```
